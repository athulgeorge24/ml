```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score,classification_report
```

```python
#load the dataset
iris=load_iris()
x=iris.data
y=iris.target
print("Feature names: ")
print(iris.feature_names)
print("\nTarget names:")
print(iris.target_names)
print("\nShape of dataset",x.shape)
```

```python
#convert to binary form
y_binary=(y!=0).astype(int)
print("Binary labels")
print(np.unique(y_binary,return_counts=True))

#orginally the dataset clases:0,1,2
#select 2 features to visualization
x_two=x[:,[2,3]]
```

```python
x_train,x_test,y_train,y_test= train_test_split(
    x_two,
    y_binary,
    test_size=0.2,
    random_state=42,
    stratify=y_binary
)
#standarize the feature
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
#trauin linear sv,
linear_svm=SVC(
    kernel='linear',
    C=1.0
)
linear_svm.fit(x_train_scaled,y_train)

#make predictions
y_pred = linear_svm.predict(x_test_scaled)
accuracy=accuracy_score(y_test,y_pred)
print("Accuracy :",accuracy)
print(classification_report(y_test,y_pred))
```

```python
#initialize boundary
plt.figure(figsize=(10,7))

#plot training points
plt.scatter(
    x_train_scaled[:,0],
    x_train_scaled[:,1],
    c=y_train,
    s=60,
    cmap='coolwarm',
    edgecolors='k'
)

#create grid
ax=plt.gca()
xlim=ax.get_xlim()
ylim=ax.get_ylim()
xx=np.linspace(xlim[0],xlim[1],200)
yy=np.linspace(ylim[0],ylim[1],200)
XX,YY=np.meshgrid(xx,yy)#it combine the dp
xy=np.vstack([XX.ravel(),YY.ravel()]).T

#calc decision fn
Z=linear_svm.decision_function(xy)
Z=Z.reshape(XX.shape)
#DRAW DECISON BOUNDARY
ax.contour(
    XX,
    YY,
    Z,
    levels=[-1,0,1],
    linestyles=['--','-','--']
)


ax.scatter(
    linear_svm.support_vectors_[:,0],
    linear_svm.support_vectors_[:,1],
    s=180,
    facecolors='none',
    edgecolors='black',
    linewidths=2,
    label='Support Vectors'
)
plt.xlabel("Petal Length (Standardized)")
plt.ylabel("Pedal Width (Standardized)")
plt.title("Linear SVM : Setosa vs Non-Setosa")
plt.legend()
plt.show()
```

```python
#use all features
x_full=iris.data
y_full=iris.target
x_train_full,x_test_full,y_train_full,y_test_full=train_test_split(
    x_full,
    y_full,
    test_size=0.2,
    random_state=42,
    stratify=y_full
)
#Sscalar the featiures
scaler_full=StandardScaler()
x_train_full=scaler_full.fit_transform(x_train_full)
x_test_full=scaler_full.transform(x_test_full)
```

```python
#create linear,ploynomial abd rbf svm models
models={
    "Linear":SVC(kernel="linear",C=1.0),
    "Polynomial":SVC(kernel="poly",degree=3,C=1.0),
    "RBF":SVC(kernel="rbf",C=1.0,gamma="scale")
}
```

```python
results={}

for name,model in models.items():
    model.fit(x_train_full,y_train_full)
    y_pred=model.predict(x_test_full)
    accuracy=accuracy_score(y_test_full,y_pred)
    results[name]=accuracy
    print(f"\n{name} Kernel")
    print("-" *30)
    print("Accuracy",accuracy)
    print(
        classification_report(
            y_test_full,
            y_pred,
            target_names=iris.target_names
        )
    )
            
```
